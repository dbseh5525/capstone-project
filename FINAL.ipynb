```python
from flask import Flask, request, send_file
from flask_cors import CORS
from pyngrok import ngrok
import io, torch, cv2
import numpy as np
from PIL import Image
from diffusers import StableDiffusionPipeline, DPMSolverMultistepScheduler
from rembg import remove, new_session

app = Flask(__name__)
CORS(app)

PROJECT_DIR = '/content/drive/MyDrive/dog_lora_project'

# ===== 모델 로드 (서버 시작 시 1회) =====
pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float16,
    safety_checker=None,
).to("cuda")
pipe.scheduler = DPMSolverMultistepScheduler.from_config(pipe.scheduler.config)
pipe.load_lora_weights(
    f"{PROJECT_DIR}/lora_my_dog/checkpoint-600",
    weight_name="pytorch_lora_weights.safetensors",
)
bg_session = new_session("u2net")

PROMPT = (
    "a photo of sks dog, white pomeranian, fluffy white fur, all white fur, "
    "full body shot, standing pose, whole body visible from head to paws, "
    "flat vector illustration, cute mascot character, chibi proportions, "
    "big round head, round chubby face, round cheeks, "
    "big round black eyes, small black nose, thick black outline, "
    "short round ears, small rounded ears close to head, "
    "simple flat shading, "
    "pure white background, plain solid background, no text, no logo, no badge, "
    "clean vector art, centered composition, high quality"
)
NEGATIVE = (
    "brown fur, tan fur, black fur, bear face, dog breed other than pomeranian, "
    "long ears, rabbit ears, pointy ears, tall ears, "
    "elongated face, narrow face, long snout, angular face, "
    "wood texture, wooden background, circular badge, sticker border, round frame, "
    "logo, text, watermark, letters, "
    "close up, head only, face only, cropped, portrait, "
    "realistic, photo, 3d render, anime girl, humanoid, personification, "
    "texture, crosshatch, pattern background, gray background, dark background, "
    "deformed, extra limbs, blurry, low quality, ugly"
)


def keep_dog_only(pil_img, erode_size=9):
    """가장 큰 덩어리(강아지)만 남기기"""
    arr = np.array(pil_img.convert("RGBA"))
    alpha = arr[..., 3]
    mask = (alpha > 30).astype(np.uint8)
    k = np.ones((erode_size, erode_size), np.uint8)
    eroded = cv2.erode(mask, k, iterations=1)
    n, labels, stats, _ = cv2.connectedComponentsWithStats(eroded, connectivity=8)
    if n > 1:
        largest = 1 + np.argmax(stats[1:, cv2.CC_STAT_AREA])
        core = (labels == largest).astype(np.uint8)
        core = cv2.dilate(core, k, iterations=1)
        final_mask = (core > 0) & (mask > 0)
        arr[~final_mask] = [0, 0, 0, 0]
    return Image.fromarray(arr)


@app.route("/generate", methods=["POST"])
def generate():
    seed = int(request.form.get("seed", 1))

    generator = torch.Generator(device="cuda").manual_seed(seed)
    img = pipe(
        prompt=PROMPT,
        negative_prompt=NEGATIVE,
        guidance_scale=6.0,
        num_inference_steps=30,
        generator=generator,
    ).images[0]

    nobg = remove(img, session=bg_session)
    final = keep_dog_only(nobg, erode_size=9)

    buf = io.BytesIO()
    final.save(buf, format="PNG")
    buf.seek(0)
    return send_file(buf, mimetype="image/png")


ngrok.set_auth_token("3IuTYphCzh7b5C45QsHeLqqnRF2_5gVfHjeKEmHecPZYTG5ct")
public_url = ngrok.connect(5000)
print("API URL:", public_url)
app.run(port=5000)
```
